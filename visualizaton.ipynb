```python
# normal distubation = most data mean tira distubrate xa
import numpy as np

# generate 500 random std marks (normally distributed around 72)

np.random.seed(42)
marks = np.random.normal(loc =72, scale=12, size = 500).clip(40,100)

# define 10-point bin boundaries ;[40,50,60,70,80,90,100]
bin_edges = range(40,105,10)

plt.figure(figsize=(9,5))

plt.hist(marks, bins = bin_edges, color = 'lightblue', edgecolor='black')
plt.title('Final exam score distribution (N = 500)')
plt.xlabel('Score range')
plt.ylabel('no of stds')
plt.xticks(bin_edges)
plt.grid(axis = 'y',linestyle = '--')
plt.show()

```

```python
import numpy as np

# generate 500 random std marks (normally distributed around 72)

np.random.seed(42)
income= np.random.lognormal(mean =8, sigma=0.5, size = 500)

# define 10-point bin boundaries ;[40,50,60,70,80,90,100]
bin_edges = np.linspace(0,15000,31)

plt.figure(figsize=(15,5))

plt.hist(income, bins = bin_edges, color = 'lightblue', edgecolor='black')
plt.title('lognormal  distribution (N = 500)')
plt.xlabel('MOntly income')
plt.ylabel('frequency')
plt.xticks(bin_edges)
plt.grid(axis = 'y',linestyle = '--')
plt.show()
```

```python
fig, axes = plt.subplots(2,2)

axes[0,0].plot(x, y)
axes[0,0].set_title('line')

axes[0,1].bar(x, y)
axes[0,1].set_title('Bar')

axes[1,0].scatter(range(5), y)
axes[1,0].set_title('scatter')

axes[1,1].hist(y)
axes[1,1].set_title('Histogram')

plt.tight_layout()
plt.show()
```

#  Seaborn

```python
df = sns.load_dataset('tips')
print(df.head())
```

#  Seaborn Scatter Plot

```python
sns.scatterplot(
    data = df,
    x = 'total_bill',
    y = 'tip'
)

plt.show()
```

```python
sns.scatterplot(
    data = df,
    x = 'total_bill',
    y = 'tip',
    hue = 'sex'
)

plt.show()
```

```python
sns.violinplot(
    data = df,
    x = 'day',
    y = 'total_bill'
)

plt.show()
```

#  Heatmap

```python
#  correlation visualization
#  first calculate correlation
#  cauzation : one variable directly influences or causes a change in another variable.

correlation = df.corr(numeric_only=True)

print(correlation)
```

```python
sns.heatmap(correlation)
plt.show()
```

```python
sns.heatmap(
correlation,
annot = True
)
plt.show()
```

# Pairplot

```python
sns.pairplot(
    df,
    hue = 'sex')

plt.show()
```
